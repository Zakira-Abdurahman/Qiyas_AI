```python
#lab12
#data wrangling 
import pandas as pd
import numpy as np
```

```python
#1 hospital patient registration, cleaning dataset
data = {
    'patient_ID': ['P001', 'P002', 'P003', 'P004', 'P005'],
    'Full_Name': ['Abebe Kebede ', ' Hana Tesfaye', ' Samuel Bekele ', ' Meron Alemu ', 'Meron Alemu'],
    'Age': [35, np.nan, 42, 29, 29],
    'Address': [
        'Bole Road, Addis Ababa, Bole, 03, H123',
        'CMC Road, Addis Ababa, Yeka, 10, H201',
        'Piassa, Addis Ababa, Arada, 01, H505',
        'Megenagna, Addis Ababa, Bole, 05, H901',
        'Megenagna, Addis Ababa, Bole, 05, H901'
    ]
}
df= pd.DataFrame(data)

print("1 original dataset:\n", df)
print("\n2 number of missing values in each column:\n" , df.isnull().sum())

#replace missing age with mean, remove duplicate rows,remove leading and trailing spaces from full name, split full name in to first and last name , split address into street, city, subcity,woreda and house_no
df["Age"] = df["Age"].fillna(df["Age"].mean())
df = df.drop_duplicates()
df["Full_Name"] = df['Full_Name'].str.strip()
df[["First_Name", "Last_Name"]] = df["Full_Name"].str.split(n=1, expand=True)
df[["Street", "City", "Subcity", "Woreda", "House_No"]] = (df["Address"].str.split(", ", expand=True))

print("8 Cleaned Dataset:\n", df)








```

```python
#2 online store sales report
data = {
    'Month': ['Jan','Jan','Feb','Feb','Mar','Mar'],
    'Product':['Laptop','Phone','Laptop','Phone','Laptop','Phone'],
    'Sales':[15,np.nan,18,20,15,25]
}
df = pd.DataFrame(data)
print("1 Original data:\n", df)
print("2 Missing Values:\n", df.isnull().sum())
print("3 After filling missing sale with 0:\n", df["Sales"].fillna(0))
print("4 Total sales:", df["Sales"].sum())
pivot = df.pivot_table(index="Month", columns="Product", values="Sales")
print("5 Pivot Table:\n", pivot)
print("6 pivot table after replacing missing value with 0:\n", df.fillna(0))
print("7 Highest sales month:", pivot.sum(axis=1).idxmax())


```

```python
#3 employee email processing , split email into username and domain, convert usernames to uppercase,create company email column ending with @company.com, count number of employees using gmail and display only employees with gmail accounts 
data = {
    'Employee_ID': ['E1','E2','E3','E4'],
    'Email': ['abebe@gmail.com','hana@gmail.com','samuel@outlook.com','meron@gmail.com']
}
df = pd.DataFrame(data)
df[["Username", "Domain"]] = df["Email"].str.split("@", expand=True)
df["Username"] = df["Username"].str.upper()
df["company_Email"] = df["Username"].str.lower() + "@company.com"
gmail_count = (df["Domain"] == "gmail.com").sum()
print("4 Number of gmail users:", gmail_count)
print("\n5 employees with gmail accounts:\n", df[df["Domain"] == "gmail.com"])



```

```python
#4 bank loan approval dashboard, create a count column, create pivot table as region as rows and loan type and status as columns, replace missing values with zero, calculate total number of approved loans, find region with highest number of approved loans 

data = {
'Region':['Addis','Addis','Adama','Adama','Hawassa','Bahir Dar'],
'Loan_Type':['Personal','Mortgage','Personal','Mortgage','Personal','Mortgage'],
'Status':['Approved','Rejected','Approved','Approved','Rejected','Approved']
}

df = pd.DataFrame(data)

df["count"] = 1
pivot = df.pivot_table(
    index="Region",
    columns=["Loan_Type", "Status"],
    values="count",
    aggfunc="sum",
    fill_value=0
)
print("2 and 3 Pivot Table:\n", pivot)

print("4 Total approved loans:", df[df["Status"] == "Approved"].shape[0])

print("5 Region with highest approved loans:", df[df["Status"] == "Approved"].groupby("Region").size().idxmax())
```

```python
# Telecom Customer Usage Analysis

import pandas as pd
import numpy as np

data = {
    'Customer_ID':['C1','C1','C2','C2','C1','C3'],
    'Month':['Jan','Jan','Jan','Jan','Feb','Feb'],
    'Service':['Data','Voice','Data','Voice','Data','Voice'],
    'Usage_GB':[12,5,8,10,np.nan,15]
}

df = pd.DataFrame(data)

print("1 Missing values:\n", df.isnull().sum())

# 2 fill missing Usage_GB with mean
df["Usage_GB"] = df["Usage_GB"].fillna(df["Usage_GB"].mean())

# 3 pivot table
pivot = df.pivot_table(
    index="Customer_ID",
    columns=["Month", "Service"],
    values="Usage_GB",
    aggfunc="sum",
    fill_value=0
)

print("3 Pivot table:\n", pivot)

pivot = pivot.fillna(0)

total_usage = df.groupby("Customer_ID")["Usage_GB"].sum()
print("5 Total usage:\n", total_usage)
print("5 Highest usage customer:", total_usage.idxmax())

#6average usage per customer
avg_usage = df.groupby("Customer_ID")["Usage_GB"].mean()
print("6 Average usage:\n", avg_usage)
```

```python
#6 product code cleaning and analysis 

data = {
    'Product_Code': ['TV-2025-001','LAP-2025-017', 'PHN-2024-023','TAB-2025-011' ],
    'Price': ['45000', '65000', '32000', '55000']
}

df = pd.DataFrame(data)

# 1convert Price into integer
df["Price"] = df["Price"].astype(int)
# 2split Product_Code
df[["Category", "Year", "Product_Number"]] = df["Product_Code"].str.split("-", expand=True)
print("3 Average price:", df["Price"].mean())
print("4 Most expensive product:", df.loc[df["Price"].idxmax(), "Product_Code"])
print("5 Products released in 2025\n",df[df["Year"] == "2025"])
print("6 Total sales:", df["Price"].sum())
```

```python
#7 university student information cleaning

data = {
    'Student_ID': ['S1','S2','S3','S3'],
    'Full_Name': ['Abel Bekele', ' Hana Alemu ', 'Meron Tadesse', 'Meron Tadesse'],
    'Department': ['CS','IT',None,'IT'],
    'CGPA': [3.6,3.9,3.2,3.2]
}

df = pd.DataFrame(data)


print("1 Display dataset\n",df)
print("2 Missing values:\n", df.isnull().sum())
# 3remove duplicates
df = df.drop_duplicates()
# 4 Remove spaces
df["Full_Name"] = df["Full_Name"].str.strip()
# 5 split name
df[["First_Name", "Last_Name"]] = df["Full_Name"].str.split(" ", expand=True)

# 6fill missing department
df["Department"] = df["Department"].fillna("Unknown")

print("7 Highest CGPA student:\n", df.loc[df["CGPA"].idxmax()])

avg = df["CGPA"].mean()
print("8 Average CGPA:", avg)
print("9 Above average students:\n", df[df["CGPA"] > avg])
```

```python
#8 human resource attendance analysis
data = {
    'Employee_ID': ['E1','E1','E1','E2','E2','E3','E3'],
    'Month': ['Jan','Jan','Feb','Jan','Feb','Jan','Feb'],
    'Status': ['Present','Absent','Present','Present','Present','Absent','Absent']
}

df = pd.DataFrame(data)

# 1 Count column
df["Count"] = 1

# 2pivot table
pivot = df.pivot_table(
    index="Employee_ID",
    columns=["Month", "Status"],
    values="Count",
    aggfunc="sum",
    fill_value=0
)


pivot = pivot.fillna(0)
print(pivot)

df_absent = df[df["Status"] == "Absent"].groupby("Employee_ID").size()
print("4 Total absences:\n", df_absent)


print("5 Max absences:", df_absent.idxmax())


total_records = df.groupby("Employee_ID").size()
print("6 Total records:\n", total_records)


absence_pct = (df_absent / total_records) * 100
print("7 Absence %:\n", absence_pct)

best = absence_pct.idxmin()
print("8 Best attendance:", best)
```

```python
# challenge exercise: customer orders analysis
data = {
    'Customer_ID': ['C001','C002','C003','C003'],
    'Full_Name': ['Abebe Kebede', ' Hana Alemu ', 'Meron Tadesse', 'Meron Tadesse'],
    'Address': [
        'Bole Road, Addis Ababa, Bole, 03, H125',
        'CMC Road, Addis Ababa, Yeka, 10, H201',
        'Piassa, Addis Ababa, Arada, 01, H505',
        'Piassa, Addis Ababa, Arada, 01, H505'],
    'Product': ['Laptop','Phone','Laptop','Tablet'],
    'Amount': [45000, np.nan, 32000, 28000]
}

df = pd.DataFrame(data)
print( "1 missing values:\n", df.isnull().sum())
# 2fill missing Amount with average
df["Amount"] = df["Amount"].fillna(df["Amount"].mean())
df = df.drop_duplicates()
df["Full_Name"] = df["Full_Name"].str.strip()
df[["First_Name", "Last_Name"]] = df["Full_Name"].str.split(" ", expand=True)
df[["Street", "City", "Subcity", "Woreda", "House_No"]] = df["Address"].str.split(", ", expand=True)
df["Count"] = 1
pivot = df.pivot_table(
    index="Customer_ID",
    columns="Product",
    values="Count",
    aggfunc="sum",
    fill_value=0
)
print(pivot)

print("9 Customer who spent most money:",df.groupby("Customer_ID")["Amount"].sum().idxmax())
avg_amount = df["Amount"].mean()
print("10 Average purchase amount:", avg_amount)

cust_total = df.groupby("Customer_ID")["Amount"].sum()
print("11 customers above average purchase:", cust_total[cust_total > avg_amount])
#12 create Full_Address
df["Full_Address"] = (
    df["Street"] + ", " +
    df["City"] + ", " +
    df["Subcity"] + ", " +
    df["Woreda"] + ", " +
    df["House_No"]
)
#13remove intermediate address columns
df = df.drop(columns=["Street","City","Subcity","Woreda","House_No"])

print("14 Final dataset:\n", df)







```
