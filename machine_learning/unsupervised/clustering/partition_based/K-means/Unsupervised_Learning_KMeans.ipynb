Question 1: Clustering Principles

A valid clustering should satisfy the following properties:
Internal Homogeneity
External Separation
Exhaustive Partition

For each property:
1. Explain what it means.
2. Give one real-world example.
3. Explain what would happen if that property were violated.else

Property 1: Internal Homogeneity
1. meaning:Points inside the same cluster should be similar to each other or close together.
2. example: like music app grouping songs. If one cluster is romantic songs then all songs inside should sound romantic.
3. If violated: If the cluster mixes romantic songs with others, the group loses meaning you can’t describe it with one clear label anymore.

Property 2 :External Separation
1. meaning: different clusters should be clearly different from each other.
2. example: in a shopping mall, one cluster could be teen shoppers and another senior shoppers. Their buying habits should be distinct.
3. if violated: If the clusters teen and senior shoppers mixed together the algorithm can’t tell them apart, and predictions or recommendations will be confusing.

Property 3 :Exhaustive Partition
 1. meaning: every data point must belong to some cluster no one is left out.
 2. example: in fraud detection, every transaction must be either in the normal cluster or the suspicious cluster.
 3. If violated: if some transactions don’t belong anywhere, the system ignores them. That could mean missing fraud cases or losing valuable insights.

Question 4: Distance Metric Properties

1. Non‑negativity → Distances are always ≥ 0; e.g., dist((2,3),(5,7))=5; ensures no negative similarity.

2. Identity → Distance between a point and itself = 0; e.g., dist((4,6),(4,6))=0; guarantees identical points are recognized.

3. Symmetry → Distance(A,B)=Distance(B,A); e.g., dist((2,3),(5,7))=5 both ways; keeps clustering fair.

4. Triangle Inequality → Dist(A,C) ≤ Dist(A,B)+Dist(B,C); e.g., A=(0,0),B=(3,0),C=(3,4): 5 ≤ 3+4; ensures geometric consistency.

### Question 19: Compare the three distance measures

| Property            | Euclidean Distance                                | Manhattan Distance                                | Chebyshev Distance                              |
|---------------------|---------------------------------------------------|--------------------------------------------------|------------------------------------------------|
| Formula             | √((x₂−x₁)² + (y₂−y₁)²)                           | |x₂−x₁| + |y₂−y₁|                                | max(|x₂−x₁|, |y₂−y₁|)                          |
| Interpretation      | Straight‑line (shortest path)                     | Grid/block path (sum of steps)                   | Maximum coordinate difference (chess king move) |
| Typical Applications| GPS, geometry, clustering                         | City navigation, robotics, text similarity       | Chess, scheduling, max‑distance problems        |
| Advantages          | True geometric distance, intuitive                | Simple, works well on grids, robust to outliers  | Captures worst‑case difference, easy to compute |
| Limitations         | Sensitive to scaling, harder in high dimensions   | Ignores diagonal shortcuts, can overestimate     | Less intuitive, ignores overall path length     |
