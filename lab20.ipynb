```python
#part2 data Preparation, Label Encoding Categorical Variables

le = LabelEncoder()
categorical_cols = [
    'city',
    'store_name',
    'manufacturer',
    'brand',
    'class',
    'size',
    'sku',
    'price_bracket'
]

for col in categorical_cols:
    df[col + '_encoded'] = le.fit_transform(df[col])

print("Encoding completed successfully!")
df.head()

```
