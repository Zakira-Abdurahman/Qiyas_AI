```python

#lab13
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
```

```python
# metplotlib 
# 1 monthly sales trend 
months = ['Jan', 'Feb', 'Mar', 'Apr','May']
sales = [120, 150, 170, 160, 190]
plt.plot(months, sales, color='green', linestyle='--', marker='s', markersize=8, label='Sales')
plt.title('Monthly Sales')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.legend()
plt.show()
```

```python
# 2 students scores 
subjects = ['Math', 'Physics', 'Chemistry', 'Biology']
scores = [85, 78, 92, 88]
plt.plot(subjects, scores, color='red', marker='*', linewidth=3, markerfacecolor='yellow', label='Scores')
plt.title('Exam Scores')
plt.xlabel('Subjects')
plt.ylabel('Scores')
plt.legend()
plt.show()
```

```python
# 3 company revenue
years = [2021, 2022, 2023, 2024]
revenue = [500, 700, 850, 1000]
plt.plot(years, revenue, color='blue', linestyle=':', marker='d', label='Revenue')
plt.title('Company Revenue')
plt.xlabel('Years')
plt.ylabel('Revenue')
plt.legend()
plt.show()
```

```python
# 4 fruit supply (vertical bar chart)
fruits = ['Apple', 'Orange', 'Banana', 'Mango']
counts = [50, 70, 40, 60]
colors = ['red', 'orange', 'yellow', 'green']

fig, ax = plt.subplots()
ax.bar(fruits, counts, color=colors, label='Fruits')
ax.set_title('Fruit Supply')
ax.set_ylabel('Quantity')
ax.legend()
plt.show()

```

```python
#Class exercise: find highest using barchart, check it
sales = pd.DataFrame({
    "day": ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"],
    "cups_sold": [120, 135, 110, 145, 160, 210, 190]
})

colors = ["blue"] * len(sales)
colors[sales["cups_sold"].idxmax()] = "red"

plt.bar(sales["day"], sales["cups_sold"], color=colors)
plt.title("Daily Cups Sold")
plt.show()
```

```python
#5 stacked bar chart (students by gender)
departments = ('CS', 'IS', 'SE', 'IT')
students = {
    'Male': np.array([70, 65, 80, 50]),
    'Female': np.array([60, 55, 70, 45])
}

fig, ax = plt.subplots()

male = ax.bar(departments, students['Male'], width=0.7, label='Male')
female = ax.bar(
    departments,
    students['Female'],
    width=0.7,
    bottom=students['Male'],
    label='Female'
)

ax.bar_label(male)
ax.bar_label(female)
ax.set_title('Students by Department and Gender')
ax.legend()

plt.show()


```

```python
# bonus challenge: build a dashboard
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
sales = [120, 150, 170, 160, 190]

fruits = ['Apple', 'Orange', 'Banana', 'Mango']
counts = [50, 70, 40, 60]
colors = ['red', 'orange', 'yellow', 'green']

movies = ['Avatar', 'Titanic', 'Inception', 'Frozen']
ratings = [8.5, 7.9, 9.0, 8.2]

departments = ['CS', 'IS', 'SE', 'IT']
male = np.array([70, 65, 80, 50])
female = np.array([60, 55, 70, 45])

# Create dashboard
fig, ax = plt.subplots(2, 2, figsize=(10, 8))

#1monthly sales trend
ax[0, 0].plot(months, sales, marker='o', color='green', label='Sales')
ax[0, 0].set_title("Monthly Sales Trend")
ax[0, 0].legend()

#2 Fruit Supply
bars = ax[0, 1].bar(fruits, counts, color=colors, label='Quantity')
ax[0, 1].bar_label(bars)
ax[0, 1].set_title("Fruit Supply")
ax[0, 1].legend()

#3Movie Ratings
ax[1, 0].barh(movies, ratings, color='skyblue')
ax[1, 0].set_title("Movie Ratings")

#4student Distribution by Gender
b1 = ax[1, 1].bar(departments, male, label='Male')
b2 = ax[1, 1].bar(departments, female, bottom=male, label='Female')

ax[1, 1].bar_label(b1)
ax[1, 1].bar_label(b2)
ax[1, 1].set_title("Student Distribution by Gender")
ax[1, 1].legend()

plt.tight_layout()
plt.show()
```
