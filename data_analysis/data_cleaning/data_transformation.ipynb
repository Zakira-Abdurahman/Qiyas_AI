```python
data = { 
    'Student_ID': ['S001', 'S002', 'S003', 'S004', 'S005'], 
    'Mathematics' : [85, 75, 95, 65, 90], 
    'Science' : [78, 88, 82, 85, 79], 
    'English' : [90, 85, 80, 70, 95], 
    'IT' : [89, 65, 88, 92, 98] 
} 
result = pd.DataFrame(data) 
result
```

```python
result["Average"] = result[["Mathematics", "Science", "English", "IT"]].mean(axis=1)
result
```

```python
result['Weight_Average'] = (
    result['Mathematics'] * 0.4 +
    result['Science'] * 0.2 +
    result['English'] * 0.2 +
    result['IT'] * 0.2
)

result
```

```python
def standardize(column):
    mean = np.mean(column)
    std_dev = np.std(column)
    return (column - mean) / std_dev

result['Math_std'] = standardize(result['Mathematics'])
result['sci_std'] = standardize(result['Science'])
result['Eng_std'] = standardize(result['English'])
result['IT_std'] = standardize(result['IT'])
result
```

```python
def assign_grade(weighted_avg):
    if weighted_avg >= 85:
        return 'A'
    elif weighted_avg >= 75:
        return 'B'
    elif weighted_avg >= 65:
        return 'C'
    elif weighted_avg >= 55:
        return 'D'
    else:
        return 'F'

result["Final_Grade"] = result["Weight_Average"].apply(assign_grade)
result
```
