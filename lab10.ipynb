```python
import pandas as pd 
import numpy as np

```

```python

#exercise1 removing duplicate records
#identify duplicate rows, count duplicate rows,remove duplicates,keep only last occurrence
df = pd.DataFrame({
    "student_id": [101,102,103,102,104,101],
    "name": ["Abel", "Sara","John", "Sara","Helen","Abel"],
    "Score":[80,90,75,90,88,80]
})
print(df.duplicated())
print("Duplicate Count:", df.duplicated().sum())
df_clean = df.drop_duplicates()
df_last = df.drop_duplicates(keep="last")
print(df_clean)
print(df_last)
```

```python
#exercise2 duplicate emails
#find duplicate eamils and remove duplicate emails
df = pd.DataFrame({
    "id": [1,2,3,4],
    "email": ["a@gmail.com","b@gmail.com","a@gmail.com","c@gmail.com"]
})

duplicates = df[df.duplicated(subset=["email"])]
print(duplicates)
clean_df = df.drop_duplicates(subset=["email"])
print(clean_df)
```

```python
#exercise3 standardize city names
#convert to uppercase, lowercase,title case
df = pd.DataFrame({
    "city": ["addis ababa","ADDIS ABABA","Addis Ababa","adDis abAba"]
})

df["upper"] = df["city"].str.upper()
df["lower"] = df["city"].str.lower()
df["title"] = df["city"].str.title()
print(df)
```

```python
#exercise4 remove extra spaces
df = pd.DataFrame({
    "name":[" Abel","Sara "," John"," Helen"]
})
df["name"] = df["name"].str.strip()
print(df)
```

```python

#6 kg to pounds
df = pd.DataFrame({
    "weight_kg": [55,70,85,100]
})

df["weight_lb"] = (df["weight_kg"] * 2.20462).round(2)

print(df)
```

```python
#7 celsius to fahrenheit

df = pd.DataFrame({"temp_c":[0,25,37,100]})
df["temp_f"] = (df["temp_c"] * 9/5) + 32
print(df)
```

```python
#8handling missing values missing values,fill missing ages with mean, fill names with unknown
df = pd.DataFrame({
    "name": ["Abel","Sara",None,"John"],
    "age": [20,np.nan,25,30]
})

print(df.isnull())
print(df.isnull().sum())
df["age"] = df["age"].fillna(df["age"].mean())
df["name"] = df["name"].fillna("Unknown")
print(df)

```

```python
#9 correect data entry errors , standardize gender values
df = pd.DataFrame({
    "gender":["Male","male","MALE","Female","female"]
})

df["gender"] = df["gender"].str.title()
print(df)
```

```python
#10 currency conversion, convert usd to etb,assume 1 usd = 135etb
df = pd.DataFrame({
    "salary_usd":[500,800,1200,1500]
})
exchange_rate = 135
df["salary_etb"] = df["salary_usd"] * exchange_rate
print(df)
```

```python
#11 outlier detection, calculate q1 and q2, calculate IQR, identify outliers,remove outliers
df = pd.DataFrame({
    "salary":[3500,4000,4200,3900,4100,45000]
})
Q1 = df["salary"].quantile(0.25)
Q3 = df["salary"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = df[
   (df["salary"] < lower) | 
   (df["salary"] > upper)]
print(outliers)

clean_df = df[
    (df["salary"] >= lower) &
    (df["salary"] <= upper)]
print(clean_df)
```
