```python
import pandas as pd 
import numpy as np

```

```python

#exercise1 removing duplicate records
#identify duplicate rows, count duplicate rows,remove duplicates,keep only last occurrence
df = pd.DataFrame({
    "student_id": [101,102,103,102,104,101],
    "name": ["Abel", "Sara","John", "Sara","Helen","Abel"],
    "Score":[80,90,75,90,88,80]
})
print(df.duplicated())
print("Duplicate Count:", df.duplicated().sum())
df_clean = df.drop_duplicates()
df_last = df.drop_duplicates(keep="last")
print(df_clean)
print(df_last)
```

```python
#exercise2 duplicate emails
#find duplicate eamils and remove duplicate emails
df = pd.DataFrame({
    "id": [1,2,3,4],
    "email": ["a@gmail.com","b@gmail.com","a@gmail.com","c@gmail.com"]
})

duplicates = df[df.duplicated(subset=["email"])]
print(duplicates)
clean_df = df.drop_duplicates(subset=["email"])
print(clean_df)
```

```python
#exercise3 standardize city names
#convert to uppercase, lowercase,title case
df = pd.DataFrame({
    "city": ["addis ababa","ADDIS ABABA","Addis Ababa","adDis abAba"]
})

df["upper"] = df["city"].str.upper()
df["lower"] = df["city"].str.lower()
df["title"] = df["city"].str.title()
print(df)
```

```python
#exercise4 remove extra spaces
df = pd.DataFrame({
    "name":[" Abel","Sara "," John"," Helen"]
})
df["name"] = df["name"].str.strip()
print(df)
```

```python
# standardize dates, conver to datatime, standardize to yyyy-mm-dd, extract year and extract month
df = pd.DataFrame({
    "date":["2026-01-05","05/02/2026","March 10,2026","2026.04.15"]
})
df["date"] = pd.to_datetime(df["date"], format="mixed")
df["formatted"] = df["date"].dt.strftime("%Y-%m-%d")
df["year"] = df["date"].dt.year
df["month"] = df["date"].dt.month
print(df)
```

```python

#6 kg to pounds
df = pd.DataFrame({
    "weight_kg": [55,70,85,100]
})

df["weight_lb"] = (df["weight_kg"] * 2.20462).round(2)

print(df)
```

```python
#7 celsius to fahrenheit

df = pd.DataFrame({"temp_c":[0,25,37,100]})
df["temp_f"] = (df["temp_c"] * 9/5) + 32
print(df)
```

```python
#8handling missing values missing values,fill missing ages with mean, fill names with unknown
df = pd.DataFrame({
    "name": ["Abel","Sara",None,"John"],
    "age": [20,np.nan,25,30]
})

print(df.isnull())
print(df.isnull().sum())
df["age"] = df["age"].fillna(df["age"].mean())
df["name"] = df["name"].fillna("Unknown")
print(df)

```

```python
#9 correect data entry errors , standardize gender values
df = pd.DataFrame({
    "gender":["Male","male","MALE","Female","female"]
})

df["gender"] = df["gender"].str.title()
print(df)
```

```python
#10 currency conversion, convert usd to etb,assume 1 usd = 135etb
df = pd.DataFrame({
    "salary_usd":[500,800,1200,1500]
})
exchange_rate = 135
df["salary_etb"] = df["salary_usd"] * exchange_rate
print(df)
```

```python
#11 outlier detection, calculate q1 and q2, calculate IQR, identify outliers,remove outliers
df = pd.DataFrame({
    "salary":[3500,4000,4200,3900,4100,45000]
})
Q1 = df["salary"].quantile(0.25)
Q3 = df["salary"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = df[
   (df["salary"] < lower) | 
   (df["salary"] > upper)]
print(outliers)

clean_df = df[
    (df["salary"] >= lower) &
    (df["salary"] <= upper)]
print(clean_df)
```

```python
#12 phone number standardization, remove symbol, convert all numbers to +251 format
df = pd.DataFrame({
    "phone": [
        "0911223344",
        "+251911223344",
        "251911223344",
        "0911-223-344"
    ]
})

df["phone"] = df["phone"].str.replace("-","", regex=False)
df["phone"] = df["phone"].str.replace("+","",regex=False)
df["phone"] = df["phone"].apply(
    lambda x: "+251" + x[1:] if x.startswith("0") else "+251" + x[3:]
)
print(df)

```

```python
#13 email cleaning, remove spaces,convert to lowercase and validate emails
df = pd.DataFrame({
    "email": [" ABEL@GMAIL.COM ","sara@gmail.com","JOHN@YAHOO.COM"]
})

df["email"] = df["email"].str.strip()
df["email"] = df["email"].str.lower()
df["valid"] = df["email"].str.contains("@")
print(df)
```

```python
#14 product name normalization, standardize product name and compare unique counts before and after cleaning
df = pd.DataFrame({
    "product": ["Laptop","laptop","LAPTOP","Laptop"]
})
print("Before:", df["product"].nunique())
df["product"] = (
    df["product"]
    .str.strip()
    .str.lower()
)
print("After:", df["product"].nunique())
print(df)
```

```python
#15 mixed data cleaning, clean names,fill missing values,standardize dates,detect outliers
df = pd.DataFrame({
    "name": [" Abel ", "Sara",None,"JOHN"],
    "age": [20,np.nan,25,30],
    "salary": [4000,4500,5000,4500],
    "join_date": ["2026-01-01","01/02/2026","March 10, 2026","2026.04.15"]
})

df["name"] = (df["name"].fillna("Unknown").str.strip().str.title())
df["age"] = df["age"].fillna(df["age"].mean())
df["join_date"] = pd.to_datetime(df["join_date"], format="mixed")
Q1 = df["salary"].quantile(0.25)
Q3 = df["salary"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = df[(df["salary"] < lower) |(df["salary"] > upper)]
print(df)
print("Outliers")
print(outliers)
```
