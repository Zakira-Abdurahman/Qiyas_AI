```python

# label Encoding + no Scaling + linear Regression 1
# 1 Label Encoding
categorical_columns = [
    "city",
    "store_name",
    "manufacturer",
    "brand",
    "class",
    "size",
    "sku",
    "price_bracket"
]

le = LabelEncoder()
for column in categorical_columns:
    df[column] = le.fit_transform(df[column])


#2 features and Target
X = df.drop(columns=["value_sales"])
y = df["value_sales"]


#3split Dataset
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42
)

print("Training Shape :", X_train.shape)
print("Testing Shape  :", X_test.shape)


#4 Train linear regression
linear_model = LinearRegression()
linear_model.fit(X_train, y_train)


#5model equation
print("Linear Regression Equation")
print(f"Intercept : {linear_model.intercept_:.4f}")

coef_df = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": linear_model.coef_
})

coef_df["Absolute Coefficient"] = coef_df["Coefficient"].abs()
coef_df = coef_df.sort_values(
    by="Absolute Coefficient",
    ascending=False
)

print("\nFeature Coefficients")
print(coef_df)


#6 predictions
y_pred_train = linear_model.predict(X_train)
y_pred_test = linear_model.predict(X_test)


# 7Evaluation

train_rmse = np.sqrt(
    mean_squared_error(y_train, y_pred_train)
)

test_rmse = np.sqrt(
    mean_squared_error(y_test, y_pred_test)
)

train_r2 = r2_score(
    y_train,
    y_pred_train
)

test_r2 = r2_score(
    y_test,
    y_pred_test
)


# 8results
print("Experiment 1 Results")
print(f"Training RMSE : {train_rmse:.2f}")
print(f"Testing RMSE  : {test_rmse:.2f}")
print(f"Training R²   : {train_r2:.4f}")
print(f"Testing R²    : {test_r2:.4f}")

#9 save
results = pd.DataFrame({
    "Model": ["Linear Regression"],
    "Encoding": ["Label Encoding"],
    "Scaling": ["None"],
    "Train RMSE": [train_rmse],
    "Test RMSE": [test_rmse],
    "Train R2": [train_r2],
    "Test R2": [test_r2]
})

print("\nComparison Table")
print(results)
```

```python

# label Encoding + StandardScaler + Linear Regression 2

categorical_columns = [
    "city",
    "store_name",
    "manufacturer",
    "brand",
    "class",
    "size",
    "sku",
    "price_bracket"
]

le = LabelEncoder()
for column in categorical_columns:
    df[column] = le.fit_transform(df[column])




X = df.drop(columns=["value_sales"])
y = df["value_sales"]


X_train, X_test, y_train, y_test = train_test_split(
    X,  y,
    test_size=0.20,
    random_state=42
)

# standard Scaling
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)


#train Model
linear_model = LinearRegression()
linear_model.fit(X_train_scaled, y_train)


#coefficients
coef_df = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": linear_model.coef_
})

coef_df["Absolute Coefficient"] = coef_df["Coefficient"].abs()
coef_df = coef_df.sort_values(
    by="Absolute Coefficient",
    ascending=False
)

print(coef_df)


#prediction
y_pred_train = linear_model.predict(X_train_scaled)
y_pred_test = linear_model.predict(X_test_scaled)

#evaluation
train_rmse = np.sqrt(
    mean_squared_error(y_train, y_pred_train)
)

test_rmse = np.sqrt(
    mean_squared_error(y_test, y_pred_test)
)

train_r2 = r2_score(
    y_train,
    y_pred_train
)

test_r2 = r2_score(
    y_test,
    y_pred_test
)

#result
print("Experiment 2")
print("Label Encoding + StandardScaler + Linear Regression")
print(f"Training RMSE : {train_rmse:.2f}")
print(f"Testing RMSE  : {test_rmse:.2f}")
print(f"Training R²   : {train_r2:.4f}")
print(f"Testing R²    : {test_r2:.4f}")


#save
results = pd.DataFrame({
    "Model":["Linear Regression"],
    "Encoding":["Label Encoding"],
    "Scaling":["StandardScaler"],
    "Train RMSE":[train_rmse],
    "Test RMSE":[test_rmse],
    "Train R2":[train_r2],
    "Test R2":[test_r2]
})

print(results)
```
