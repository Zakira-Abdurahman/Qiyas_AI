```python
import pandas as pd
```

```python
#dataset1: customers
custumers = pd.DataFrame({
    "Customer_ID": ["C001", "C002", "C003", "C004", "C005"],
    "Custumer_Name": ["Abel", "Sara", "John", "Helen", "Samuel"],
    "City": ["Addis Abeba", "Adama", "Jimma", "Bahir Dar", "Hawassa"]
})

#dataset2: orders
orders = pd.DataFrame({
    "Order_ID": [101,102,103,104,105,106],
    "Customer_ID":["C001","C002","C003","C001","C004","C006"],
    "Product":["Laptop","Phone","Keyboard","Mouse","Monitor","Tablet"],
    "Amount":[1200,800,150,45,400,700]

})

#dataset3 : employee salary 
employees = pd.DataFrame({
   "Employee_ID":[1,2,3,4,5,6],
   "Department":["IT","IT","Sales","Sales","HR","HR"],
   "Employee":["Abel","Sara","John","Helen","Samuel","David"],
   "Salary":[5000,6500,4500,4700,5200,4900]
})
```

```python
#partI
# #exercise 1
inner_join = pd.merge(custumers, orders, on='Customer_ID', how='inner')
print("Inner Join:\n",  inner_join)
print("\n number of rows returned:", len(inner_join))
print("missing customer is C005 and C006")
```

```python
#exercise 2 
left_join =  pd.merge(custumers, orders, on='Customer_ID', how='left')
print("Left Join:\n" , left_join)

#Which customer information is missing?
print("missing customer imformation : C005(Samuel)")
#Why does it happen?
print("Because Samuel exists in the Customers but has no matching record in the Orders dataset")
```

```python
#exercise3 
right_join= pd.merge(custumers, orders, on='Customer_ID', how='right')
print("Right Join:\n" , right_join)
print("customer who has never placed an order: C005 (Samuel)")
print("missing values in right join: Custumer_Name and City for Customer_ID C006 ")
```

```python
#exercise4
outer_join= pd.merge(custumers, orders, on='Customer_ID', how='outer')
print("outer Join:\n" , outer_join)
print("\nTotal rows returned:", len(outer_join))
print("Total rows returned: 7")
print("customer only in Customers dataset: C005 (Samuel)")
print("customer only in Orders dataset: C006")
```

```python
#exercise5 
# Left Join
left_join = pd.merge(orders, custumers, on="Customer_ID", how="left")
print("before replacing\n",left_join)
left_join = left_join.fillna("Unknown")
print("\n after replacing missing values with unknown\n",left_join)

```

```python
#partII filtering data
#exercise6
# average male and female customers order amount 
# city with the highest and least order 

merged = pd.merge(orders, custumers, on="Customer_ID")
male = merged[merged["Custumer_Name"].isin(["Abel", "John", "Samuel", "David"])]
print("average male customers order amount:", male["Amount"].mean())
```
