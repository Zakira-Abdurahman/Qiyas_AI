```python
import pandas as pd
```

```python
#dataset1: customers
custumers = pd.DataFrame({
    "Customer_ID": ["C001", "C002", "C003", "C004", "C005"],
    "Custumer_Name": ["Abel", "Sara", "John", "Helen", "Samuel"],
    "City": ["Addis Abeba", "Adama", "Jimma", "Bahir Dar", "Hawassa"]
})

#dataset2: orders
orders = pd.DataFrame({
    "Order_ID": [101,102,103,104,105,106],
    "Customer_ID":["C001","C002","C003","C001","C004","C006"],
    "Product":["Laptop","Phone","Keyboard","Mouse","Monitor","Tablet"],
    "Amount":[1200,800,150,45,400,700]

})

#dataset3 : employee salary 
employees = pd.DataFrame({
   "Employee_ID":[1,2,3,4,5,6],
   "Department":["IT","IT","Sales","Sales","HR","HR"],
   "Employee":["Abel","Sara","John","Helen","Samuel","David"],
   "Salary":[5000,6500,4500,4700,5200,4900]
})
```

```python
#partI
# #exercise 1
inner_join = pd.merge(custumers, orders, on='Customer_ID', how='inner')
print("Inner Join:\n",  inner_join)
print("\n number of rows returned:", len(inner_join))
print("missing customer is C005 and C006")
```

```python
#exercise 2 
left_join =  pd.merge(custumers, orders, on='Customer_ID', how='left')
print("Left Join:\n" , left_join)

#Which customer information is missing?
print("missing customer imformation : C005(Samuel)")
#Why does it happen?
print("Because Samuel exists in the Customers but has no matching record in the Orders dataset")
```

```python
#exercise3 
right_join= pd.merge(custumers, orders, on='Customer_ID', how='right')
print("Right Join:\n" , right_join)
print("customer who has never placed an order: C005 (Samuel)")
print("missing values in right join: Custumer_Name and City for Customer_ID C006 ")
```

```python
#exercise4
outer_join= pd.merge(custumers, orders, on='Customer_ID', how='outer')
print("outer Join:\n" , outer_join)
print("\nTotal rows returned:", len(outer_join))
print("Total rows returned: 7")
print("customer only in Customers dataset: C005 (Samuel)")
print("customer only in Orders dataset: C006")
```

```python
#exercise5 
# Left Join
left_join = pd.merge(orders, custumers, on="Customer_ID", how="left")
print("before replacing\n",left_join)
left_join = left_join.fillna("Unknown")
print("\n after replacing missing values with unknown\n",left_join)

```

```python
# average male and female customers order amount 
# city with the highest and least order 
custumers["Gender"] = ["Male", "Female", "Male", "Female", "Male"]
merged = pd.merge(custumers, orders, on="Customer_ID", how="inner")
avg_gender = merged.groupby("Gender")["Amount"].mean()

print("Average Order Amount by Gender:\n", avg_gender)

city_sales = merged.groupby("City")["Amount"].sum()
print("City with Highest Orders:", city_sales.idxmax())

print("City with Lowest Orders:", city_sales.idxmin())
```

```python
#partVI apply()
#exercise 28
def salary_level(Salary):
    if Salary >= 6000:
        return 'High'
    elif Salary >= 5000:
        return 'Medium'
    else:
        return 'Low'
 
employees['salary_level'] = employees['Salary'].apply(salary_level)
print(employees)


    


```

```python
#exercise 29
def department_type(dep):
    if dep == 'IT':
        return 'Technical'
    else:
        return 'Non-Technical'
    
employees['department_type'] = employees['Department'].apply(department_type)
print(employees)

```

```python
# partVII challenge exercises
# exercise 30
merged = pd.merge(custumers, orders, on='Customer_ID', how='inner')
result = merged[['Custumer_Name', 'Product', 'Amount']].sort_values(by='Amount', ascending=False)
print("sorted customer purchases high to low:\n", result)
```

```python
# exercise 31
top_customer = merged.loc[merged["Amount"].idxmax()]
print("customer who spend the most:\n", top_customer)
```

```python
# exercise 32 
result = merged.groupby("Custumer_Name")["Amount"].sum().sort_values(ascending=False)
print("Total amount spent by each customer High to Low:\n", result)

```

```python
#exercise 33
result = merged.groupby("Custumer_Name")["Amount"].sum()
result = result[result > 1000]

print("Customers with total purchases > 1000:\n", result)

```

```python
#exercise 34
def order_category(amount):
    if amount >= 1000:
        return "Premium"
    elif amount >= 500:
        return "Standard"
    else:
        return "Basic"

merged["Order_Category"] = merged["Amount"].apply(order_category)

print(merged[["Custumer_Name", "Product", "Amount", "Order_Category"]])
```

```python
#Exercise 35 (Mini Project)
merged = pd.merge(custumers, orders, on="Customer_ID", how="outer")
print("1 Total Sales:", orders["Amount"].sum())
print("2 Average Sales:", orders["Amount"].mean())
print("3Highest Sale:", orders["Amount"].max())
print("4 Lowest Sale:", orders["Amount"].min())
print("5 Number of Customers:", custumers["Customer_ID"].nunique())

top_customer = merged.groupby("Custumer_Name")["Amount"].sum().idxmax()
print("6Top Customer:", top_customer)

no_orders = merged[merged["Amount"].isna()][["Custumer_Name", "City"]]
print("7 Customers Without Orders:\n", no_orders)

unknown_orders = merged[merged["Custumer_Name"].isna()][["Customer_ID", "Product", "Amount"]]
print("8Orders Without Registered Customers:\n", unknown_orders)

sales_per_customer = merged.groupby("Custumer_Name")["Amount"].sum()
print("9 Total Sales Per Customer:\n", sales_per_customer)

top3 = sales_per_customer.sort_values(ascending=False).head(3)
print("10Top 3 Customers:\n", top3)


```
