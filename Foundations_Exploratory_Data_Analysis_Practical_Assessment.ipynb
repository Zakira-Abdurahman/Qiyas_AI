```python
# 5 Multiple Conditions given studnet mark 80–100 → A, 70–79 → B,60–69 → C,Below 60 → F, display grade

mark = float(input("Enter the student's mark: "))
if mark >= 80 and mark <= 100:
    print("Grade: A")
elif mark >= 70:
    print("Grade: B")
elif mark >= 60:
    print("Grade: C")
else:
    print("Grade: F")
```

```python
#6 for loop: Print numbers from 1 to 20 and display only the even numbers
for i in range(1, 21):
    if i % 2 == 0:
        print(i)
```

```python
#7While Loop:Use a while loop to calculate the sum of numbers from 1 to 100
i = 1
total = 0
while i <= 100:
    total = total + i
    i = i + 1

print("Sum from 1 to 100 is:",total)
```

```python
#9 Lambda Function:Create a lambda function that squares a number and test it on values from 1 to 10
square = lambda x: x * x
for i in range(1, 11):
    print("Square of", i, "is", square(i))
```

```python
#10 List Comprehension: using list comprehension, generate a list containing squares from 1 to 20
squares = [x**2 for x in range(1, 21)]
print(squares)
```

```python
#13 Statistics Using NumPy, For an array of exam scores find mean,max,min,sd
scores = np.array([40,60,55,77,43])
print("Mean:", np.mean(scores))
print("Maximum Value:", np.max(scores))
print("Minimum Value:", np.min(scores))
print("Standard Deviation:", np.std(scores))

```

```python
#14 pandas dataframes, create dataframe containing student name, age,score and display first 3 rows
data = {
    "Name": ["Zak", "Sara", "Yad", "Baba"],
    "Age": [20,25,30,35],
    "Score": [90,80,70,50]
}
df= pd.DataFrame(data)
print(df.head(3))
```

```python
#15Selecting Rows and Columns: Using a DataFrame, Display only the Name column and Display students whose score is greater than 70
data = {
    "Name": ["Zak", "Sara", "Yad", "Baba"],
    "Age": [20,25,30,35],
    "Score": [90,80,70,50]
}
df= pd.DataFrame(data)
print(df["Name"])
print(df[df["Score"] > 70])
```

```python
#16 Adding New Columns, create column called status and assign pass for scores>= 50 and fail otherwise
data = {
    "Name": ["Zak", "Sara", "Yad", "Baba"],
    "Age": [20,25,30,35],
    "Score": [29,80,70,50]
}
df = pd.DataFrame(data)

df["Status"] = df["Score"].apply(
    lambda x: "Pass" if x >= 50 else "Fail"
)
print(df)
```
