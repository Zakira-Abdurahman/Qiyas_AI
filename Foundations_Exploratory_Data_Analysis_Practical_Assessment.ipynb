```python
#section E : integrated data wrangling
#23 cleaning customer dataset, create dataframe containing missing values,duplicate rows and inconsistent text and clean the dataset and display cleaned version
df = pd.DataFrame({
    "Customer": ["zAk", "BABA", None, "Sara", "zak"],
    "City": ["addis ababa", "ADDIS ABABA", "AdDis abaBa", None, "addis ababa"],
    "Age": [22, 30, None, 25, 22]
})
print(df)
df["Customer"] = df["Customer"].fillna("Unknown")
df["City"] = df["City"].fillna("Unknown")
df["Age"] = df["Age"].fillna(df["Age"].mean())

df["Customer"] = df["Customer"].str.title()
df["City"] = df["City"].str.title()
df = df.drop_duplicates()

print("Cleaned Dataset:", df)


```

```python
#24 outlier detectiion using IQR 
df = pd.DataFrame({
    "sales": [21, 25, 230, 30, 19, -5, 27]
})

#calculate q1 , q3, IQR
Q1 = df["sales"].quantile(0.25)
Q3 = df["sales"].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

#identify outliers
outliers = df[(df["sales"] < lower_bound) | (df["sales"] > upper_bound)]
print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Outlier Values:", outliers["sales"].values)
print(outliers)





```

```python
#25 outlier handling and complete data cleaning, create dataset with missing values, duplicate rows,incorrect text format ,extreme outlier values and then
# detect missing values, fill missing values, remove duplicate,standardize text values,detect outliers using IQR,removes or caps the outliers and display final cleaned dataset

df = pd.DataFrame({
    "name": [" zak ", "Sara", None, "JOHN", "zak"],
    "age": [20, np.nan, 25, 30, 20],
    "salary": [4000, 4500, 50000, 5000, 4000],
    "city": ["addis ababa", "ADDIS ABABA", "AdDis abaBa", None, "addis ababa"]
})
print(df)

print("\n1 missing values:",df.isnull().sum())

#2 fill missing values
df["name"] = df["name"].fillna("Unknown")
df["city"] = df["city"].fillna("Unknown")
df["age"] = df["age"].fillna(df["age"].mean())
print("\n2 after filling missing values:", df)

# 3Standardize text values
df["name"] = df["name"].str.strip().str.title()
df["city"] = df["city"].str.strip().str.title()
print("\n3 after Standardizing text:", df)

#4remove duplicate rows
df = df.drop_duplicates()
print("\n4 removed duplicate rows:", df)

#5Detect outliers using IQR
Q1 = df["salary"].quantile(0.25)
Q3 = df["salary"].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = df[(df["salary"] < lower) | (df["salary"] > upper)]
print("\n5 outliers detected:",outliers)

#6cap outliers
df["salary"] = np.where(df["salary"] < lower, lower, np.where(df["salary"] > upper, upper, df["salary"]))
print("\n6 capd outliers:", df)
#7Final cleaned dataset
print("\n7 final cleaned dataset:",df)







```
